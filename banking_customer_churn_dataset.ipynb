```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 1700

provinces = [
    "Gauteng",
    "Western Cape",
    "Eastern Cape",
    "KwaZulu-Natal",
    "Free State",
    "Limpopo",
    "Mpumalanga",
    "North West",
    "Northern Cape"
]

account_types = [
    "Savings",
    "Current",
    "Business",
    "Student"
]

genders = [
    "Male",
    "Female"
]

data = []

for i in range(1, n + 1):

    age = np.random.randint(18, 70)
    tenure = np.random.randint(1, 21)

    monthly_transactions = np.random.randint(1, 80)

    monthly_balance = round(
        np.random.uniform(500, 250000), 2
    )

    credit_score = np.random.randint(300, 851)

    churn_status = np.random.choice(
        ["Active", "Churned"],
        p=[0.80, 0.20]
    )

    data.append([
        i,
        age,
        np.random.choice(genders),
        np.random.choice(provinces),
        np.random.choice(account_types),
        tenure,
        monthly_transactions,
        monthly_balance,
        credit_score,
        churn_status
    ])

df = pd.DataFrame(
    data,
    columns=[
        "customer_id",
        "age",
        "gender",
        "province",
        "account_type",
        "tenure_years",
        "monthly_transactions",
        "monthly_balance",
        "credit_score",
        "churn_status"
    ]
)

df.head()
```

```python
df.to_csv(
    "south_african_banking_customer_churn.csv",
    index=False
)

print("Dataset saved successfully!")
```
